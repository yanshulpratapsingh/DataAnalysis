# 1. Introducing Arrays

## 1.1 NumPy Arrays

```python
# NumPy arrays
python_list = [3, 2, 5, 8, 4, 9, 7, 6, 1]
array = np.array(python_list)
print(array)
print(type(array))
```

## 1.2 Creating 2D Arrays from Lists

```python
python_list_of_lists = [[3, 2, 5],
                       [9, 7, 1],
                       [4, 3, 6]]
array_2d = np.array(python_list_of_lists)
print(array_2d)
```

## 1.3 Python Lists vs NumPy Arrays

| Python List                                | NumPy Array                             |
| ------------------------------------------ | --------------------------------------- |
| Can naturally contain different data types | Normally uses one common `dtype`        |
| `type()` → `list`                          | `type()` → `numpy.ndarray`              |
| General-purpose collection                 | Designed for numerical/array operations |
| Example: `[1, "Hello", 3.5]`               | Example: `np.array([1, 2, 3])`          |


```python
python_list = ["beep", False, 56, 0.945, [3, 2, 5]]
print(python_list)

numpy_boolean_array = np.array([[True, False], [True, True], [False, True]])
numpy_float_array = np.array([1.9, 5.4, 8.8, 3.6, 3.2])
print(numpy_boolean_array)
print(numpy_float_array)
```

## 1.4 Creating Arrays from Scratch

```python
zeros_array = np.zeros((5, 3))
random_array = np.random.random((2, 4))
arange_array = np.arange(-3, 4)

print(zeros_array)
print(random_array)
print(arange_array)
```

## 1.5 np.arange()

np.arange() is used to create a NumPy array containing evenly spaced values within a given range.

```python
print(np.arange(4))
print(np.arange(-3, 4, 3))
print(np.arange(-3, 4))
```

# 2. Array Dimensionality

## 2.1 3D Arrays

```python
array_1_2D = np.array([[1, 2], [5, 7]])
array_2_2D = np.array([[8, 9], [5, 7]])
array_3_2D = np.array([[1, 2], [5, 7]])

array_3D = np.array([array_1_2D, array_2_2D, array_3_2D])
print(array_3D)
print("Shape:", array_3D.shape)
```

## 2.2 Vector, Matrix and Tensor Arrays

| Type       | Dimensions | Example                          | Shape     |
| ---------- | ---------: | -------------------------------- | --------- |
| **Vector** |         1D | `[1, 2, 3]`                      | `(3,)`    |
| **Matrix** |         2D | `[[1,2],[3,4]]`                  | `(2,2)`   |
| **Tensor** | 3D or more | `[[[1,2],[3,4]], [[5,6],[7,8]]]` | `(2,2,2)` |


```python
vector = np.array([1, 2, 3])
matrix = np.array([[1, 2], [3, 4]])
tensor = np.zeros((2, 2, 2))

print("Vector:", vector)
print("Matrix:\n", matrix)
print("Tensor shape:\n", tensor)
```

# 8. Adding and Removing Data

## 8.1 Concatenating Rows

```python
classroom_ids_and_sizes = np.array([[1, 22], [2, 21], [3, 27], [4, 26]])
new_classrooms = np.array([[5, 30], [5, 17]])

combined = np.concatenate((classroom_ids_and_sizes, new_classrooms))
print(combined)
```

## 8.2 Concatenating Columns

```python
grade_levels_and_teachers = np.array([
    [1, "James"], [1, "George"], [3, "Amy"], [3, "Meehir"]
])

combined_columns = np.concatenate(
    (classroom_ids_and_sizes, grade_levels_and_teachers), axis=1
)
print(combined_columns)
```

## 8.3 Shape and Dimension Compatibility

```python
array_1D = np.array([1, 2, 3])
column_array_2D = array_1D.reshape((3, 1))
row_array_2D = array_1D.reshape((1, 3))

print(column_array_2D)
print(row_array_2D)
```

## 8.4 Deleting Rows

np.delete(array, index, axis=0) 0-row, 1-col

```python
classroom_data = combined_columns
print(np.delete(classroom_data, 1, axis=0))
```

## 8.5 Deleting Columns

```python
print(np.delete(classroom_data, 1, axis=1))
```

## 8.6 Deleting Without an Axis

np.delete() without specifying axis, NumPy first treats the array as a flattened 1D array and then deletes the specified element(s).

```python
print(np.delete(classroom_data, 1))
```

# 9. Summarizing Data

## 9.1 Aggregating Methods

```python
security_breaches = np.array([
    [0, 5, 1],
    [0, 2, 0],
    [1, 1, 2],
    [2, 2, 1],
    [0, 0, 0]
])

print("sum:", security_breaches.sum())
print("min:", security_breaches.min())
print("max:", security_breaches.max())
print("mean:", security_breaches.mean())
print("cumsum:\n", security_breaches.cumsum())
```

## 9.2 Summing Data

```python
print(security_breaches.sum())
```

## 9.3 Aggregating Rows

```python
print(security_breaches.sum(axis=0))
```

## 9.4 Aggregating Columns

```python
print(security_breaches.sum(axis=1))
```

## 9.5 Minimum and Maximum

```python
print(security_breaches.min())
print(security_breaches.max())
print(security_breaches.min(axis=1))
```

## 9.6 Mean

```python
print(security_breaches.mean())
print(security_breaches.mean(axis=1))
```

## 9.7 Cumulative Sums

```python
cum_sums_by_client = security_breaches.cumsum(axis=0)
print(cum_sums_by_client)
```

## 9.8 Graphing Summary Values

```python
import matplotlib.pyplot as plt

cum_sums_by_client = security_breaches.cumsum(axis=0)
plt.plot(np.arange(1, 6), cum_sums_by_client[:, 0], label="Client 1")
plt.plot(np.arange(1, 6), cum_sums_by_client.mean(axis=1), label="Average")
plt.show()
```

# 12. Saving and Loading Arrays

## 12.1 RGB Arrays

```python
rgb = np.array([
    [[255, 0, 0], [255, 0, 0], [255, 0, 0]],
    [[0, 255, 0], [0, 255, 0], [0, 255, 0]],
    [[0, 0, 255], [0, 0, 255], [0, 0, 255]]
])

plt.imshow(rgb)
plt.show()
```

## 12.2 Saving and Loading .npy Files

```python
# Save
np.save("logo.npy", rgb)

# Load
logo_rgb_array = np.load("logo.npy")
print(logo_rgb_array.shape)

plt.imshow(logo_rgb_array)
plt.show()
```

## 12.3 Examining RGB Data

Channel 0 → Red
Channel 1 → Green
Channel 2 → Blue

```python
red_array = logo_rgb_array[:, :, 0]
green_array = logo_rgb_array[:, :, 1]
blue_array = logo_rgb_array[:, :, 2]

print(red_array)
print(green_array)
print(blue_array)
```

## 12.4 Updating RGB Data

```python
dark_logo_array = np.where(logo_rgb_array == 255, 50, logo_rgb_array)

plt.imshow(dark_logo_array)
plt.show()
```

## 12.5 Saving Arrays as .npy Files

```python
np.save("dark_logo.npy", dark_logo_array)

# Load it again
loaded_dark_logo = np.load("dark_logo.npy")
print(loaded_dark_logo.shape)
```

# 15. Stacking and Splitting

## 15.1 Slicing Dimensions

```python
rgb = np.array([
    [[255, 0, 0], [255, 255, 0], [255, 255, 255]],
    [[255, 0, 255], [0, 255, 0], [0, 255, 255]],
    [[0, 0, 0], [0, 255, 255], [0, 0, 255]]
])

red_array = rgb[:, :, 0]
green_array = rgb[:, :, 1]
blue_array = rgb[:, :, 2]

print(red_array)
```

## 15.2 Splitting Arrays

```python
red_array, green_array, blue_array = np.split(rgb, 3, axis=2)

print(red_array)
print("Shape:", red_array.shape)
```

## 15.3 Trailing Dimensions

```python
red_array_2D = red_array.reshape((3, 3))

print(red_array_2D)
print("Shape:", red_array_2D.shape)
```

## 15.4 Array Division Rules

```python
# This is intentionally invalid because the third dimension
# has size 3 and cannot be split equally into 5 parts.
# np.split(rgb, 5, axis=2)
```

## 15.5 Plotting 2D Image Data

```python
red_array, green_array, blue_array = np.split(logo_rgb_array, 3, axis=2)

plt.imshow(red_array)
plt.show()
```

## 15.6 Stacking 2D Arrays

```python
# Create three compatible 2D channels
red_2d = np.zeros((3, 3), dtype=np.int32)
green_2d = rgb[:, :, 1]
blue_2d = rgb[:, :, 2]

stacked_rgb = np.stack([red_2d, green_2d, blue_2d], axis=2)

print("Shape:", stacked_rgb.shape)
plt.imshow(stacked_rgb)
plt.show()
```
