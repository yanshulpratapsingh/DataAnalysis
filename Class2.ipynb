```python
# Perform a task to calculate The Simple Interest

principal = int(input("Enter Principal Amount : "))
rate = int(input("Enter Rate : "))
time = int(input("Enter Time : "))

si = (principal * rate * time) / 100

print("Simple Interest : ",si)

# Compound Interest

ci = si - principal

print("Compound Interest : ",ci)
```

```python
# Factorial of a No.

n = int(input("Enter the Number : "))
fact = 1

for i in range(1, n+1):
  fact = fact * i

print(fact)
```

```python
# Reverse a number

n = int(input("Enter the Number : "))
rev = 0

while n > 0:
  rem = n % 10
  rev = rev * 10 + rem
  n = n // 10

print(rev)
```

```python
# Count a digit

n = int(input("Enter the Number : "))
count = 0

while n > 0:
  n = n // 10
  count = count + 1

print(count)
```

```python
# Sum of a Digit

n = int(input("Enter the Number : "))
sum = 0

while n > 0:
  rem = n % 10
  sum = sum + rem
  n = n // 10

print(sum)
```
