```python
#27
travel_df = pd.DataFrame({
    "Destination": ["Agra", "Delhi", "Jaipur", "Mumbai", "Goa"],
    "Days": [2, 3, 4, 5, 6]
})

# np.arange(5) add karna
travel_df["Code"] = np.arange(5)
display(travel_df)

# Sirf index 1 aur 4 par values wali Series se replace karna
travel_df["Code"] = pd.Series([10, 40], index=[1, 4])
display(travel_df)
```

```python
#28
school_df = pd.DataFrame({
    "Student": ["Amit", "Sumit", "Rahul", "Pooja", "Neha"],
    "Class": ["10th", "10th", "12th", "11th", "12th"],
    "Section": ["A", "B", "A", "C", "B"]
})

# Missing 'House' column ke sath reindex karna
school_df = school_df.reindex(columns=["Student", "Class", "Section", "House"])

# House column mein ek common value assign karna
school_df["House"] = "Red"
display(school_df)
```

```python
#29
inventory_df = pd.DataFrame({
    "Item": ["Pen", "Pencil", "Eraser", "Notebook", "Scale"],
    "Quantity": [10, 20, 15, 30, 25]
})

# Partially indexed Series add karna (e.g., indexes 1 aur 3 par)
inventory_df["Discount"] = pd.Series([5.0, 10.0], index=[1, 3])
display(inventory_df)

# Column delete karna aur phir display karna
inventory_df = inventory_df.drop(columns=["Discount"])
display(inventory_df)
```

```python
#30
import io
data = input("Apna table paste karein:\n")
lab_df = pd.read_csv(io.StringIO(data), sep="\t")

# Sequence of statements
display(lab_df)
print("Column Names:", lab_df.columns)
display(lab_df["SingleWordCol"])      # Apne table ke single-word column ka naam yahan likhein
display(lab_df["Column With Space"])  # Apne table ke space wale column ka naam yahan likhein

smaller_lab_df = lab_df[["Col1", "Col2"]] # Apne table ke hisaab se columns dein
display(smaller_lab_df)

lab_df["NewCol"] = 100
display(lab_df)

lab_df = lab_df.drop(columns=["NewCol"])
display(lab_df)
```
