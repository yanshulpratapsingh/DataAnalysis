```python
# modification of data
df['Victim Category'] = pd.cut(df['Victim Age'], bins=[0, 13, 19, 35, 59, 100], labels=['Child', 'Teenager','Young Adult', 'Adult', 'Senior'])
df
```

```python
# Replace invalid zero

df['Police Deployed'] = df['Police Deployed'].replace(0, df['Police Deployed'].mean())
df
```

```python
import matplotlib.pyplot as plt

category_counts = df['Victim Category'].value_counts()

plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct='%1.1f%%'
)

plt.title('Victim Category Distribution')
plt.show()
```
