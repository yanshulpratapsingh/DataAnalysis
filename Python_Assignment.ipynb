Topic 8: Scope — Local vs Global Variables

```python
# Q8.1. Predict the output, then run it:

def my_abs(val):
  if val < 0:
    return 0 - val
  return val
print(val)

# Explain the error in your own words.

# val is a local variable inside the my_abs() function. It exists only while the function is running,
# so print(val) outside the function cannot access it.
# Therefore, Python raises a NameError.
```

```python
# Q8.2. Create a global variable counter = 0. Write a function show_counter() that prints counter. Call the function
# and confirm it can read the global variable.

counter = 0

def show_counter():
    print(counter)

show_counter()
```

```python
# Q8.3. Now write a function increment_counter() that tries to do counter = counter + 1 without using the global
# keyword. Run it and explain the error you get.

counter = 0

def increment_counter():
    counter = counter + 1

increment_counter()

# Because counter = counter + 1 assigns a value to counter, Python treats counter as a local variable inside the function.
# But it tries to read its value before the local variable has been assigned, so Python raises an UnboundLocalError.
```

```python
# Q8.4. Fix the function from the previous question using the global keyword so it correctly increments the global
# counter variable.

counter = 0

def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print(counter)
```
